```python
def drop_columns(df,cols):
    """
    This function removes the columns from a DataFrame.
    """
    for col in cols:
        try:
            df.drop(col,axis=1,inplace=True)
        except Exception as e:
            print(f"Bir hata oluştu : {e}")
```
