```python
files = os.listdir(path="hepsiemlak/")
```

```python
df = pd.concat([pd.read_csv(f"hepsiemlak/{file}") for file in files],ignore_index=True)
```

```python
silinecekSutunlar = ['photo-count','list-view-date','left',
       'img-wrp href', 'he-lazy-image src', 'he-lazy-image src 2',
       'list-view-header', 'branded__text', 'branded__text 2',
       'img-wrp href 2', 'he-lazy-image src 3','list-view-location']
for sutun in silinecekSutunlar:
       try:
              df.drop(sutun,axis=1,inplace=True)
       except Exception as e:
              print(f"Bir hata oluştu : {e}")
```

sadece bir sütünu yazmayı unutmuş olabiliriz bu gibi senaryolarda programın cokmesini onlemek icin try except kullanabiliriz bu sayede bir sutunu yazmamış olsak bile sonradan ekleme yapabiliriz.

axis=1 sütün silme,
kalıcı şekilde silmek icin (data frame üzerinde işlem yapmak icin) inplace=True
