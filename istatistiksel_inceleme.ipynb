## Verilerin istatistiksel olarak incelenmesi ve aykırı degerlerin ayıklanması

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Verilerin yuklenmesi

```python
df = pd.read_csv("data.csv")
```

```python
df["city"] = df["city"].astype("category")
df["district"] = df["district"].astype("category")
df["neighborhood"] = df["neighborhood"].astype("category")
```

# Numerik degiskenlerin minimum maximum ve ceyreklik degerlerinin bulunması

```python
columns = df.select_dtypes(include=[np.number]).columns
```

```python
min_values = []
max_values = []

for column in columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    min_value = Q1 - 1.5 * IQR
    max_value = Q3 + 1.5 * IQR
    min_values.append(min_value)
    max_values.append(max_value)
    print(f"Column={column} : min_value={min_value} , max_value={max_value}")
```

çıkarımlarımız : 'living_room' degerlerinde cok fazla 1 degeri var 

# Aykırı Degerlerin Temizlenmesi

```python
for i, column in enumerate(columns):
    df = df[(df[column] >= min_values[i]) & (df[column] <= max_values[i])]
```

max ve min degerlerinin arasında bulunan degerleri alıyoruz 

enumerate ile index deger cifti seklinde cıktı alabiliriz.
